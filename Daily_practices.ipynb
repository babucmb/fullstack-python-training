```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack=[]
        m={")": "(", "}": "{", "]": "["}
        for ch in s:
            if ch in m:
                top=stack.pop() if stack else "#"
                if m[ch]!=top:
                    return False
                else:
                    stack.append(ch)
        return not stack

```

Given an integer n, find the total sum of digits of all numbers from 1 to n.

Input: n = 5
Output: 15
Explanation:1+2+3+4+5=15

```python
def calc_sum(n):
    #n=5
    ts=0
    f=1
    while f<=n:
        l=n%f
        c=(n//f)%10
        h=n//(f*10)
        
        ts+=h*45*f
        
        for d in range(c):
            ts+=d*f
        ts+=c*(l+1)
        f*=10

    return ts
n=5
print(calc_sum(n))
```
