```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack=[]
        m={")": "(", "}": "{", "]": "["}
        for ch in s:
            if ch in m:
                top=stack.pop() if stack else "#"
                if m[ch]!=top:
                    return False
                else:
                    stack.append(ch)
        return not stack

```

Given an integer n, find the total sum of digits of all numbers from 1 to n.

Input: n = 5
Output: 15
Explanation:1+2+3+4+5=15

```python
def calc_sum(n):
    #n=5
    ts=0
    f=1
    while f<=n:
        l=n%f
        c=(n//f)%10
        h=n//(f*10)
        
        ts+=h*45*f
        
        for d in range(c):
            ts+=d*f
        ts+=c*(l+1)
        f*=10

    return ts
n=5
print(calc_sum(n))
```

```python
#  python
def tour(petrol, distance, n):
    start = 0
    current_petrol = 0
    required_petrol = 0

    for i in range(n):

        current_petrol += (petrol[i] - distance[i])

    if current_petrol < 0:
        required_petrol += current_petrol
        start = i + 1
        current_petrol = 0
    if current_petrol + required_petrol >= 0:
        return start
    else:
        return -1

tour([7,8,4,5,6],[5,6,3,4,1],5)
```

```python
from typing import Optional
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def isBalanced(self, root: Optional[TreeNode]) -> bool:
        def check(node):
            if not node:
                return True, 0
            left_balanced, left_height = check(node.left)
            right_balanced, right_height = check(node.right)
            
            balanced = (left_balanced and right_balanced and 
                        abs(left_height - right_height) <= 1)
            return balanced, 1 + max(left_height, right_height)
        
        return check(root)[0]

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

sol = Solution()
print(sol.isBalanced(root)) 
```

```python
from typing import List
class Solution:
    def maxArea(self, height: List[int]) -> int:
        l,r=0, len(height)-1
        ma=0
        while l<r:
            w=r-l
            h=min(height[l], height[r])
            ma=max(ma,w*h)

            if height[l]<height[r]:
                l+=1
            else:
                r-=1
        return ma
sol=Solution()
inh=[1,8,6,2,5,4,8,3,7]
print(sol.maxArea(inh))
        
```
