```python
int_dict={1:"one",2:"two",3:"three"}
print(int_dict)

float_dict={1.1:"one",2.2:"two",3.3:"three"}
print(float_dict)

str_dict={"one":1,"two":2,"three":3}
print(str_dict)

hetero_dict={1:"one",2.2:"two","three":3}
print(hetero_dict)

# Tuple as key in dictionary
tuple_dict={(1,2):"one",(2,3):"two",(3,4):"three"}
print(tuple_dict)


# key must be immutable, so we can use tuple as key in dictionary but we cannot use list and set as key in dictionary because they are mutable.
```

```python
# list as key in dictionary, but it raing the eror because list is mutable and cannot be used as a key in dictionary
list_dict={[1,2]:"one",[2,3]:"two",[3,4]:"three"}
print(list_dict)
```

```python
# Set as key in dictionary, but it raing the eror because set is mutable and cannot be used as a key in dictionary
set_dict={{1,2}:"one",{2,3}:"two",{3,4}:"three"}
print(set_dict)
```

```python
dict_comp={x:x**2 for x in range(1,6)}
print(dict_comp)
```

```python
# Basic interview Questions
# Q1. Set is mutable or immutable in Python? Explain.
# A1. Set is mutable in Python. This means that you can add or remove elements from a set after it has been created. However, the elements themselves must be immutable (e.g., numbers, strings, tuples).
# But keys are immutable in dict.

# Q2. What is the time complexity of adding an element to a dictionary in Python?
# Ans:- BC, AC=0(1), WC=O(n)

#Q3. Duplicates elements are allowed in dictionary or not? Explain.
# Ans:- Duplicates elements are not allowed in dictionary. If we try to add a duplicate key, the old value will be overwritten by the new value. This is because dictionary keys must be unique.

# Q4. Heterogeneous elements are allowed in dictionary or not? Explain.
# Ans:- Heterogeneous elements are allowed in dictionary. This means that we can have keys and values of different data types in a dictionary. For example, we can have a dictionary with integer keys and string values, or string keys and list values. This flexibility allows us to store a wide variety of data in a single dictionary.

# Q5. Dictionary is Ordered or unordered in Python? Explain.
# Ans:- Dictionary is ordered in Python 3.7 and above. This means that the order in which items are inserted into a dictionary is preserved when iterating over the dictionary. 
# In earlier versions of Python, dictionaries were unordered, meaning that the order of items was not guaranteed to be the same as the order in which they were added. However, starting from Python 3.7, dictionaries maintain insertion order as an implementation detail, and this behavior was officially declared part of the Python language specification in Python 3.8.




```

# Accessing Dictionary Items

```python
drink_items={
    "Coke": 1.5, 
    "Pepsi": 1.2, 
    "Sprite": 1.0, 
    "Fanta": 1.3}
print(drink_items)

val=drink_items["Pepsi"]
val
```

```python
val=drink_items.get("Sprite")
val
```

```python
print(drink_items.keys())
print(drink_items.values())
print(drink_items.items())
```

```python
# Nested Dictionary
nested_dict={
    "Drink1": {"Name": "Pepsi", "Price": 20},
    "Drink2": {"Name": "Sprite", "Price": 25},
    "Drink3": {"Name": "Coca-Cola", "Price": 30}
}

nested_dict["Drink1"]["Name"]
```

# Insert & Update Operation on Dictionary

```python
x_dict={"name":"Adhi", "age": 25}
print(x_dict["name"])
x_dict["age"]

# update
x_dict["age"]=26
x_dict

#Insert
x_dict['City']="Chennai"
x_dict


# Insert and Update using the update() function
# update() can insert and update mutiple items together

x_dict.update({"age":30 , 'occupation': "lover Boy"})
print(x_dict)
```
