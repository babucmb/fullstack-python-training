# Delete items from the List(remove(), pop(), clear(), del())

```python
food = ['pizza', 'burger', 'pasta', 'salad', 'tacos']
print(food)

# removing 
food.remove("pizza")
print(food)

# pop() with index
var = food.pop(1)
print(var)
# pop() without index
food.pop()
print(food)

# Del() mutiple elements 
del food[:2]
print(food)

```

```python
# clear is used to remove all the elements from the list
food = ['pizza', 'burger', 'pasta', 'salad', 'tacos']
food.clear()
food
```

```python
del food
```

# List Comprehension

```python
l1=[i for i in range(1,10)]
print(l1)

l2=[i for i in range(1,11) if i%2==0]
print(l2)

l3=[i**2 for i in range(1,11)]
print(l3)

l4=[i for i in range(1,11) if i%2==0 if i%3==0]
print(l4)

l4=[i for i in range(1,11) if i%2==0 or  i%3==0]
print(l4)

l4=[i for i in range(1,11) if i%2==0 and i%3==0]
print(l4)
print(l3==l2)
print(l3 is l2)


l5=[i**2 if i%2==0 else i+2 for i in range(1,10)]
print(l5)

l6=[i**2 if i%2==0 else i+2 for i in range(1,10) if i>5 ]
print(l6)
```

```python
l7=str(l6)
r=""
for ch in l7:

    if ch=="[" or ch== "]":
        continue
    r=ch+r
    
print(r)
```
