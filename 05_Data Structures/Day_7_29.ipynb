# Updating List Items

```python
list_Polictician= ["KCR", "KTR", "Pavan", "CBN", "Balayya"]
list_Polictician[1]="Dance with"
list_Polictician
```

```python
# Update Multiple Elements: update more elements with less
list_Polictician[3:5]=["dance with"]
list_Polictician
```

```python
# update less elements with more
list_Polictician[1:4]=["Rahul", "MOdi", "Kejriwal", "Yogi", "Mamta"]
list_Polictician
```

```python
list_Polictician= ["KCR", "KTR", "Pavan", "CBN", "Balayya"]
list_pol2=["adhi", "mahesh"]
list_Polictician[3:5]=list_pol2
print(list_Polictician)
```
