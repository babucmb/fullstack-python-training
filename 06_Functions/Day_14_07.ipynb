# Enumerate Function

--> enumerate() function add the counter (index) with list / tuple items.

```python
fruits_names=["mamidipandu", "PanasaPandu", "JamaPandu"]

for index,name in enumerate(fruits_names, start=1):  # By defualt it is start from ZERO but we can give the starting point
    print(index, name)
```

```python
# Enumerate with Tuple
tu=("Mahesh","adi",101,102, 5.0,9999)
for index, name in enumerate(tu, start=101):
    print(index, name)

```

# Zip() Function

--> zip() function is used to combine/zip mutiple list/tuple items, element-wise
--> The Zip() need equal length of list/tuple

```python
hero=["Adi", "malikarjun", "pavan kalyan", "balaji"]
heroines=["Anushka", "samantha",  "rashmika", "Kajal"]
movie=["x","y","z","h"]
price=[20,30,40,50]

pairs=zip(hero,heroines, movie, price)
print(list(pairs))
```

```python
# --> The Zip() need equal length of list/tuple, if not it will not print the last one , it will discard it
hero=["Adi", "malikarjun", "pavan kalyan", "balaji"]
heroines=["Anushka", "samantha",  "rashmika", "Kajal"]
movie=["x","y","z","h"]
price=[20,30,40]

pairs=zip(hero,heroines, movie, price)
print(list(pairs))
```

# Decorators

Def:--> A decorator is a function that take another function as input, adds extra functionality and return a modified function, without changing the original function code.



```python
def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
        print("add cherry")
        print("add chocolate")
        
    return wrapper_fun
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
decorate_fun=dec_fun(cake)
decorate_fun()
```

```python
def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
        print("add cherry") #  Extra functionality
        print("add chocolate")
        
    return wrapper_fun
@dec_fun
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```

```python
# Multiple decorator for same function

def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
    return wrapper_fun
def dec_fun2(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        othfun()
        print("add cherry") #  Extra functionality
        print("add chocolate")
        
    return wrapper_fun

@dec_fun
@dec_fun2
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```
