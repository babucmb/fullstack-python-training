# Lambda Function

Def:- A lambda function is a single line of code that is used to perform a short taks without definning a full function with def.
 --> A lambda function don't have any name. 
 --> Althoigh lambda fuction can recive arguments and process them.

```python
obj= lambda a: a+100
print(obj(10))
```

```python
obj=lambda x,y:x*y
print(obj(2,10))
```

```python
obj=lambda a,b:a+b
obj(100,200)
```

Lambda function are propularaly used inside the other function to perform some small task.

```python
def fun(p):
    return lambda a:a**p # powering ** will show that powring 

obj1=fun(10)
print(obj1(2))
print(obj1(3))
```

```python
def fun(p):
    obj1= lambda a:a**p if a%2==0 else a+p
    return obj1
obj1=fun(2)
print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
# sort the below given list of tuple w,r,t marks
st=[("p",80),("d",75),("c",78)]
#print(sorted(st))
std= sorted(st, key=lambda a:a[1])
print(std)
```

```python
st.sort(key= lambda x:x[1] )
print(st)
```

```python
st=[("p",80,5),("d",75,8),("c",78,7)]
st.sort(key=lambda x:x[2])
print(st)
```

```python
st={"p":80, "d":75,"c":78}
d=st.items() # converting dictionary to list of tuple
#print(d)
sd=sorted(d,key=lambda a:a[1])
print(sd)
```

# Recursive Functions

def :- A function call it self, directly or indirectly
--> Termination condition it MUST in recursion
--> In recursive call, we must change increment or decrement

```python
def foo(i):
    if i<1:
        return
    print(i)
    foo(i-1)
foo(6)
```
