#  Default Parameters

```python
def fun(var="Ajith"):
    print(var)
fun()
fun(100)
```

```python
def func2(x=10,y=20):
    print(x+y)
func2() # output =30
func2(100)
func2(100,200)
```

# Passing Arbitrary number of arguments using *args

```python
def fun(*args):
    for i in args:
        print(i)
fun()
fun("Ajith")
fun("Adhi")

```

# Passing arbitrary Key-value based argument using  **args

```python
def aji(**GF):
    for k in GF:
        print(k, ":", GF[k])
aji(chinki=150,minki=2500)
aji(chi=150)
aji()
```
