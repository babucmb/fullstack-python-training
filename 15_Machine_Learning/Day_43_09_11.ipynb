## 1. What is Naive Bayes?

Naive Bayes is a supervised machine learning algorithm mainly used for classification. It is based on **Bayes' theorem**. It is called "naive" because it assumes that input features are conditionally independent of each other when the class is known.

Bayes' theorem:

\[
P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}
\]

- `P(A|B)`: posterior probability
- `P(B|A)`: likelihood
- `P(A)`: prior probability
- `P(B)`: evidence

The model calculates the probability of each class and predicts the class with the highest probability.

## 2. Where is Naive Bayes used?

- Email spam detection
- Sentiment analysis
- News or document classification
- Topic classification

## 3. Types of Naive Bayes

- **Gaussian Naive Bayes:** For continuous numerical features, often modeled with a Gaussian distribution.
- **Multinomial Naive Bayes:** Commonly used for text classification with word counts or frequencies.
- **Bernoulli Naive Bayes:** Used for binary features, such as whether a word is present or absent.

## 5. Import the required libraries

If needed, install them with `%pip install numpy pandas scikit-learn`.

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 6. Create a small spam classification dataset

`ham` means a normal message and `spam` means an unwanted or promotional message. This small dataset is only for learning, not for building a production spam filter.

```python
data = {
    "message": [
        "Hey, are we meeting today",
        "Please call me when you are free",
        "Are you coming to class",
        "I will send the notes tonight",
        "Let's have lunch tomorrow",
        "Can you share the project file",
        "Congratulations you won a free prize",
        "Claim your free cash reward now",
        "You have won a lottery click here",
        "Limited offer buy now and get discount",
        "Free entry in the lucky draw",
        "Claim your reward by clicking this link",
        "Please review my assignment",
        "The meeting starts at ten",
        "Your package will arrive tomorrow",
        "Win cash now, claim your prize"
    ],
    "label": [
        "ham", "ham", "ham", "ham", "ham", "ham",
        "spam", "spam", "spam", "spam", "spam", "spam",
        "ham", "ham", "ham", "spam"
    ]
}

df = pd.DataFrame(data)
df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nClass counts:")
print(df["label"].value_counts())
```

## 7. Separate input and output

`X` contains the message text and `y` contains the label to predict.

```python
X = df["message"]
y = df["label"]

print(X.head())
print(y.head())
```

## 8. Convert text into numbers

Machine learning models need numerical input. `CountVectorizer` converts messages into word-count features. I split the data first and fit the vectorizer only on training text to avoid data leakage.

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

## 9. Train the Naive Bayes model

I am using `MultinomialNB` because this example uses word-count features.

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

## 10. Make predictions

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Message": X_test_text.values,
    "Actual": y_test.values,
    "Predicted": y_pred
})

results
```

## 11. Check model performance

Accuracy is the percentage of correct predictions. With a very small dataset, the score can change depending on the split, so it is not a reliable real-world performance estimate.

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=["ham", "spam"]))
```

## 12. Predict a new message

Try changing the message and run the cell again.

```python
new_messages = [
    "Congratulations you won a free reward",
    "Can you send me the class notes"
]

new_messages_vector = vectorizer.transform(new_messages)
predictions = model.predict(new_messages_vector)

for message, prediction in zip(new_messages, predictions):
    print(message, "->", prediction)
```

## 13. Gaussian Naive Bayes with numerical data

Gaussian Naive Bayes is suitable for numerical features. This made-up dataset uses study hours and attendance to predict whether a student passes.

```python
from sklearn.naive_bayes import GaussianNB

student_data = pd.DataFrame({
    "study_hours": [1, 2, 2.5, 3, 4, 5, 6, 7, 8, 9],
    "attendance": [40, 45, 50, 55, 60, 70, 75, 80, 90, 95],
    "result": ["Fail", "Fail", "Fail", "Fail", "Pass",
               "Pass", "Pass", "Pass", "Pass", "Pass"]
})

X_num = student_data[["study_hours", "attendance"]]
y_num = student_data["result"]

gaussian_model = GaussianNB()
gaussian_model.fit(X_num, y_num)

new_student = pd.DataFrame({
    "study_hours": [6],
    "attendance": [78]
})

print("Prediction:", gaussian_model.predict(new_student)[0])
print("Class probabilities:", gaussian_model.predict_proba(new_student))
```

## 14. Advantages and limitations

### Advantages
- Simple and easy to understand.
- Fast to train and predict.
- Works well for many text classification tasks.
- Can work with relatively small training datasets.

### Limitations
- The feature-independence assumption is often unrealistic.
- Without smoothing, unseen feature combinations can cause zero probability estimates.
- Performance depends on the dataset and task.
- Predicted probabilities are not always well calibrated.

**Laplace smoothing** adds a small count (usually 1) when estimating probabilities. It helps avoid zero probability estimates for features not observed with a class.

## 15. Technical interview questions and answers

**1. What is Naive Bayes?**  
A supervised classification algorithm based on Bayes' theorem. It predicts a class using probabilities and assumes conditional independence between features given the class.

**2. Why is it called Naive Bayes?**  
Because it makes the simplifying assumption that features are conditionally independent when the class is known.

**3. Is Naive Bayes supervised or unsupervised?**  
Supervised, because it learns from labeled training data.

**4. Is Naive Bayes used for classification or regression?**  
It is mainly used for classification.

**5. What is Bayes' theorem?**  
A probability rule that calculates the probability of an event using prior information and observed evidence.

**6. What is prior probability?**  
The probability of a class before considering the current input features.

**7. What is likelihood?**  
The probability of observing the input features given a particular class.

**8. What is posterior probability?**  
The updated probability of a class after considering the observed features.

**9. What is Multinomial Naive Bayes?**  
A variant commonly used with count-based features, such as word counts in text classification.

**10. What is Gaussian Naive Bayes?**  
A variant that models numerical features using a Gaussian distribution within each class.

**11. What is Bernoulli Naive Bayes?**  
A variant designed for binary features, such as whether a word appears in a document.

**12. What is Laplace smoothing?**  
A technique that adds a small value to feature counts to prevent zero probability estimates.

**13. Why is Naive Bayes used for spam detection?**  
It is computationally efficient and can use word-based features to estimate whether a message is spam or normal.

**14. What is the difference between Multinomial and Bernoulli Naive Bayes?**  
Multinomial uses feature counts or frequencies. Bernoulli uses binary feature values, such as present or absent.

**15. What are the advantages of Naive Bayes?**  
It is simple, fast, easy to train, and often effective for text classification.

**16. What is one limitation of Naive Bayes?**  
Its conditional-independence assumption may not match real data because features can be related.

**17. What is the difference between training and testing data?**  
Training data fits the model. Test data is kept separate to check performance on unseen examples.

**18. What is overfitting?**  
When a model learns training examples too closely and performs poorly on new data.

**19. What does `fit()` do in scikit-learn?**  
It trains the model using input features and target labels.

**20. What does `predict()` do?**  
It uses the trained model to predict labels for new input data.
