# Decision Tree

```python
# Decision tree is a Supervised ML Algorithm
# Decision tree is mainly used for classification problems
# Decision tree can also be used for Regression based problems.
# Decision tree algorithm take time while training
# Decision tree is fast while making prediction.

# Features:- Age, Weight, Smoker, Target: Health issue
```

```python
## LIbraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine
```

```python
wine=load_wine()
wine
```

```python
print(dir(wine))
```

```python
x=wine.data
y=wine.target
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dt=DecisionTreeClassifier(max_depth=12) # max_depth helping in Decision Tree pruning
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)
```

```python
# ploting the DT
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(dt, feature_names=wine.feature_names,class_names=wine.target_names,filled=True)
plt.title("Decision tree for Wine Dataset")
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import fetch_california_housing
```

```python
house=fetch_california_housing()
house
```

```python
x=house.data
y=house.target
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dtt=DecisionTreeRegressor(max_depth=2) # max_depth helping in Decision Tree pruning
dtt.fit(x_train,y_train)
```

```python
dtt.score(x_test,y_test)
```

```python
# ploting the DT
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(dtt, feature_names=house.feature_names,class_names=house.target_names,filled=True)
plt.title("Decision tree for Wine Dataset")
```
