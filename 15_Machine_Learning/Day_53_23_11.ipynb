# DBSCAN

```python
# DBSCAN overcome the drawbacks if KMeans
# Kmeans connot handle spherical data. DBScan cn ahndle the spherical data.
# Kmeanss fails on spherical data because of cemtroid approach
# DBSCAN work on spherical data strcutre data becuse it works on the principal of nearest data points
# KMeans cannot handle the outliers. Becuase it will make the cluster with outliers also
# DBScan can handle the outliers. by making restriction on minimum of the data points needed to make cluster.
```

```python
# KMeans and DBScan both are unsupervised ML Algorithm.
```

```python
# Libraries
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles # to generate spherical data
```

```python
# Generate synthestic data
x,y=make_circles(n_samples=2000, factor=0.4, noise=0.05, random_state=42)
plt.scatter(x[:,0],x[:,1])

```

```python
# train our model
k = 2
kmean_model = KMeans(n_clusters = k, random_state=42)
y_kmeans = kmean_model.fit_predict(x)
y_kmeans
```

```python
plt.scatter(x[:,0], x[:,1], c = y_kmeans)
```

```python
# DBScan
db=DBSCAN(eps=0.2, min_samples=10)
y_kmeans = db.fit_predict(x)
plt.scatter(x[:,0],x[:,1],c=y_kmeans)
```
