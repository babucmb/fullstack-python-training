```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import linear_model
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv("insurance_logistic_dataset_10k.csv")
df.bought_insurance.unique() # to know the unique values in the column
df
```

```python
plt.scatter(df['age'], df['bought_insurance'], color='blue')

# The space of Logistic Regression is 'S'
```

```python
x=df.drop('bought_insurance',  axis=1)# axis=1 means we are taking the column, axis=0 means we are taking the row
y=df.bought_insurance

```

```python
x_test,x_train,y_test,y_train=train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
# train the model using Logistic Regression
log_reg_obj=LogisticRegression()
log_reg_obj.fit(x_train,y_train)
```

```python
log_reg_obj.predict([[500]]) # to predict the value of 20 years old person, whether he will buy insurance or not
```

```python
log_reg_obj.score(x_test,y_test) # to know the accuracy of the model
```

```python
# Actual vs predicted values
# where the actual test output is stored
actual_values=y_test[0:10]
predicted_values=log_reg_obj.predict(x_test[0:10])
for actual, predicted in zip(actual_values, predicted_values):
    print(f"Actual: {actual}, Predicted: {predicted}")
```
