# Confusion Matrix

```python
# Libraries
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
```

```python
# Load the inbuilt dataset
cancer_obj = load_breast_cancer()
cancer_obj
```

```python
x = cancer_obj.data
y =  cancer_obj.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.4,random_state = 42)
```

```python
# Scaling the data
scaler = StandardScaler()
x_train_Scaled = scaler.fit_transform(x_train)
x_test_Scaled = scaler.fit_transform(x_test)
```

```python
# import diff ML algo
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

```python
lreg = LogisticRegression()
svmc = SVC()
nav_guas = GaussianNB()
knn_c = KNeighborsClassifier()
dt_c = DecisionTreeClassifier()
```

```python
lreg.fit(x_train_Scaled, y_train)
svmc.fit(x_train_Scaled, y_train)
nav_guas.fit(x_train_Scaled, y_train)
knn_c.fit(x_train_Scaled, y_train)
dt_c.fit(x_train_Scaled, y_train)
```

```python
# Score 
print(lreg.score(x_test_Scaled, y_test))

print(svmc.score(x_test_Scaled, y_test))

print(nav_guas.score(x_test_Scaled, y_test))
print(knn_c.score(x_test_Scaled, y_test))
print(dt_c.score(x_test_Scaled, y_test))
```

```python
# Predic
# y_test is the target
pred_lreg = lreg.predict(x_test_Scaled)
pred_svmc= svmc.predict(x_test_Scaled)
pred_nav_guas = nav_guas.predict(x_test_Scaled)
pred_knn_c = knn_c.predict(x_test_Scaled)
pred_dt_c = dt_c.predict(x_test_Scaled)
```

```python
# y_test is the "actual" target value
import seaborn as sns
import matplotlib.pyplot as plt

# Construct the Confusion matrix

plt.figure(figsize =(5,2))
cm_lreg = confusion_matrix(y_test, pred_lreg)
sns.heatmap(cm_lreg, annot = True)
plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
```

# Accuracy Matrix

```python
# total correctly predicted / total_predicted
# Do not use accuracy matrix if data is Inbalanced
# ex : no 50/50, not suitable to not to USE when 99 % data is there
```

```python
confusion_matrix(y_test, pred_dt_c )
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, pred_dt_c ).ravel()

accuracy = (TP + TN) / (TP + TN + FN + FP )
print(accuracy )
```

```python
# Inbuilt Calculation
from sklearn.metrics import accuracy_score

arc = accuracy_score(y_test,  pred_dt_c)
arc
```

# Precison

```python
# how much acurate you are

*****# Precision = TP / (TP + FP )
# USE Precision when we concern about Type 1 ERR
# Bigger the FP -> the Precision is 0...
# Ex : spam / not Spam                                                                                                                                   iuyy
```

```python
# Using Formulae
TN, FP, FN, TP = confusion_matrix(y_test, pred_dt_c ).ravel() # ravel helps in producing 2D to 1D
 
precision = TP  / (TP + FP )
print(precision )
```

```python
# Inbuilt
from sklearn.metrics import precision_score
p_score = precision_score(y_test,  pred_dt_c)
p_score
```

# Recall / Sensitivity

```python
# Out of actual positive cals how many are predicted correctly.
# Recall = TP / (TP + FN)
# USED When concern about Tyoe 2
# Bigger the FN -> the Recall is 0...
# Ex : Covid / NO Covid
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, pred_dt_c ).ravel() # ravel helps in producing 2D to 1D
 
Recall = TP  / (TP + FN )
print(Recall )
```

```python
# Inbuilt
from sklearn.metrics import recall_score
r_score = recall_score(y_test,  pred_dt_c)
r_score
```

# Specificity & FPR (False Positive Rate)

```python
# Out of the actual negative values how many are predicted correctly
# Specificity = TN / (TN +  FP)
# There is NO Specificity for Specificity and FPR in sklearn
# Ex : Good Credit Score / Not
# Providing Bad Credit Score loan like FP
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, pred_dt_c ).ravel() # ravel helps in producing 2D to 1D
 
Specificity =  TN / (TN + FP )
print(Specificity)
```

```python
FPR = 1 - Specificity # FPR = FP / (TN + TP)
print(FPR)
```

```python
# Inbuilt
from sklearn.metrics import specificity_score
Specificity = specificity_score(y_test,  pred_dt_c)
specificity_score
```

# F1 Score

```python
# Used when concerned about both Type Errs
# F1 = 2 * [recall * precision / (Recall + precision)]
```

```python
TN, FP, FN, TP = confusion_matrix(y_test, pred_dt_c ).ravel() # ravel helps in producing 2D to 1D
 
F1 = 2 * (precision *  Recall / (precision + Recall))
print(F1)
```

```python
# Inbuilt
from sklearn.metrics import f1_score
f1 = f1_score(y_test,  pred_dt_c)
f1
```
