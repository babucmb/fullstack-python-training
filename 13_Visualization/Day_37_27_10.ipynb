# Scatter Plot

```python
# Use:- To find the relationship between two varibles
# e.g finding the relationship between age and salary


# type of relationship:
# 1. Postive relation (both variable values incerasing or both var value decreasing)
# 2. Negative relation (one increase other decrease)
# 3. No relation
import matplotlib.pyplot as plt
age=[5,8,10,20,25,30,35,40,45,50,55,60]
salary=[2,5,10,15,20,25,30,35,40,45,100,200,]
# print(age,salary)
salary.reverse()
print(salary)
plt.scatter(age,salary,s=100,color="Black")
plt.title("Age VS Salary")
plt.xlabel("Age")
plt.ylabel("salary")
```

# Pie Chart

```python
# Use:- Used to represent Less number categorical data
# To represent the data in from of precentage/ proportion

mobile_company=["Realme", "Samsung", "Nokia","Lava","Apple","IQOO","ONEPlue"]
market_share=[15,25,2,3,30,5,20]
plt.pie(market_share, labels=mobile_company, autopct='%1.1f%%')
plt.title("market_share")
#plt.show()
plt.savefig("ds20_piechart.jpg")
plt.savefig("ds20_piechart.png")
plt.savefig("ds20_piechart.pdf")
```

```python
mobile_company=["Realme", "Samsung", "Nokia","Lava","Apple","IQOO","ONEPlue"]
market_share=[15,25,2,3,30,5,20]
plt.Figure(figsize=(5,4))
plt.pie(market_share, labels=mobile_company, autopct='%1.1f%%', wedgeprops=dict(width=0.60))

plt.text(0,0,"Sales",ha="center")
plt.show()
```

# Box Plot

```python
# Use:- Used for handling outliers
# outlier: the data which is different from rest of majority data
import numpy as np
salary=[2,3,4,5.1,7.8,3.4,2.4,8.6,15,17]
np.mean(salary)
```

```python
plt.Figure(figsize=(3,2))
plt.boxplot(salary)
plt.show
```
