# Bar chart

```python
# use: used to shows the disrete values of categorial data

food_items = ["idli", "dosa", "vada", "sambar"]
like_count = [4,7,16.8,9]

plt.figure(figsize = (4,5))

plt.bar(food_items, like_count, color =["green", "black", "red","blue"])

for i, count in enumerate(like_count):
    plt.text(i, count+0.2, count, ha = "center")

plt.xlabel("food items")
plt.ylabel("Number of likes")
```

```python
# use: used to shows the disrete values of categorial data

food_items = ["idli", "dosa", "vada", "sambar"]
like_count = [4,7,16,9]

plt.figure(figsize = (4,5))

plt.bar(food_items, like_count, color =["green", "black", "red","blue"])
```

# Histogram

# Histogram works on continues data 
# Histogram is popularly used to find the shape of the data

import numpy as np
data=np.random.randn(1000)
plt.figure(figsize=(3,2)) 
plt.hist(data, edgecolor='black',bins=30)
plt.title("Histogram")
plt.xlabel("X-data")
plt.ylabel("Y-data")
