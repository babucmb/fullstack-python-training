# Method-1: using plt.subplot(): this is good for less number of plots

```python
# plt.subplot(nrow,ncols,index)

mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQOO", "ONEPlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]
plt.Figure(figsize=(6,4))

# Change Subplot to subplot
plt.subplot(2, 2, 1)  
plt.pie(market_share, labels=mobile_company)

plt.subplot(2, 2, 2)  
plt.pie(market_share, labels=mobile_company)

plt.subplot(2, 3, 5)  
plt.pie(market_share, labels=mobile_company)

plt.subplot(2, 3, 6)  
plt.pie(market_share, labels=mobile_company)

#plt.show()

plt.tight_layout() # use this is subplots are overlapping
plt.show()
```

# Different subplots  

```python
# plt.subplot(nrow,ncols,index)

mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQOO", "ONEPlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

age=[5,8,10,20,25,30,35,40,45,50,55,60]
salary=[2,5,10,15,20,25,30,35,40,45,100,200,]

food_items = ["idli", "dosa", "vada", "sambar"]
like_count = [4,7,16.8,9]
plt.Figure(figsize=(6,4))

# Change Subplot to subplot
plt.subplot(2, 2, 1)  
plt.pie(market_share, labels=mobile_company)

plt.subplot(2, 2, 2)  
plt.pie(market_share, labels=mobile_company, autopct='%1.1f%%', wedgeprops=dict(width=0.60))

plt.subplot(2, 3, 5)  
plt.scatter(age,salary,s=100,color="Black")

plt.subplot(2, 3, 6)  
plt.boxplot(salary)

plt.subplot(3,3,7)
plt.bar(food_items, like_count, color =["green", "black", "red","blue"])

#plt.show()

plt.tight_layout() # use this is subplots are overlapping
plt.show()
```

```python
# plt.subplot(nrows, ncols, index)
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()


mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

food_items = ["Idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

plt.figure(figsize = (6,4))

plt.subplot(2,3,1)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,2)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,3)
plt.imshow(digits.images[4])

plt.subplot(2,3,4)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,5)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,6)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.tight_layout()  # use this if subplots are overlapping
plt.show()
```

# Method-1: using plt.subplots(): This is good for larger numbers of subplots

```python
Dinner_dream_thali=["Biryani", "Chicken", "Rusgulla", "Roti", "Lassi"]
prec=[30,20,10,20,20]

_, axes= plt.subplots(2,2,figsize=(10,10))  # (nrows, ncols, full image size)
titles=["Gopi","Hari","Ajay","Dinesh"]
print(axes)
axes_id=axes.ravel()
print(axes_id)


for titles, ax in zip(titles,axes.ravel()):
    ax.pie(prec, labels=Dinner_dream_thali, autopct='%1.1f%%')
    ax.set_title(titles)

plt.tight_layout()
plt.show()
```

# 3D Charts in Matplotlib

```python
import numpy as np
z=np.linspace(0,10,100)
x=np.sin(z)
y=np.cos(z)
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.scatter(x,y,z)
#print(z)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the coordinates
z = np.linspace(0, 10, 100)
x = np.sin(z)
y = np.cos(z)

# 2. Add the 4th dimension (e.g., a function of z, or any data array)
w = z ** 2 

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 3. Map the 4th dimension 'w' to the color argument 'c'
scat = ax.scatter(x, y, z, c=w, cmap='viridis', s=40)

# 4. Add a color bar to act as the 4th axis indicator
cbar = fig.colorbar(scat, ax=ax, pad=0.1)
cbar.set_label('4th Dimension (W)')

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

plt.show()
```
