## 1. What is NumPy?

**Definition:** NumPy (Numerical Python) is an open-source Python library that provides support for multidimensional arrays and matrices, along with mathematical functions that operate on them.

### Interview question: What is the difference between a Python list and a NumPy array?

**Answer:**
- Python lists are built into Python; NumPy arrays are provided by the NumPy library.
- NumPy arrays are designed for efficient numerical operations, especially on large arrays.
- Lists can contain mixed types; NumPy arrays generally use one common data type.
- NumPy supports vectorized operations, so we can perform element-wise calculations without writing an explicit loop.
- Python lists can grow dynamically. NumPy arrays have a fixed shape; use array operations such as `append` (which returns a new array) or create a new array when a different size is needed.

**Interview tip:** Avoid saying NumPy is always faster. It is often faster for large numerical workloads, but performance depends on the operation and data size.

````python
# Compare a Python list and a NumPy array
import numpy as np

python_list = [10, 20, 30]
numpy_array = np.array([10, 20, 30])

print("List:", python_list)
print("Array:", numpy_array)
print("List type:", type(python_list))
print("Array type:", type(numpy_array))
print("NumPy version:", np.__version__)
````

## 2. Creating NumPy arrays

Use `np.array()` to create an array from a list.

````python
count_food_items = np.array([10, 20, 30, 40, 50])

print(count_food_items)
print(type(count_food_items))
print("Dimensions:", count_food_items.ndim)
print("Shape:", count_food_items.shape)
print("Size:", count_food_items.size)
print("Data type:", count_food_items.dtype)
````

### What do these properties mean?

- `ndim`: number of dimensions.
- `shape`: length along each dimension.
- `size`: total number of elements.
- `dtype`: data type of the array elements.

## 3. One-dimensional and two-dimensional arrays

````python
one_d = np.array([1, 2, 3, 4])
two_d = np.array([[1, 2, 3], [4, 5, 6]])

print("1-D array:\n", one_d)
print("1-D ndim:", one_d.ndim)
print("2-D array:\n", two_d)
print("2-D ndim:", two_d.ndim)
print("2-D shape:", two_d.shape)
````

## 4. Creating arrays with NumPy helpers

````python
print("Zeros:", np.zeros(4))
print("Ones:", np.ones((2, 3)))
print("Range:", np.arange(0, 10, 2))
print("Evenly spaced values:", np.linspace(0, 1, 5))
````

## 5. Indexing and slicing

NumPy indexing starts at `0`. Negative indices count from the end.

````python
arr = np.array([10, 20, 30, 40, 50])

print("First:", arr[0])
print("Last:", arr[-1])
print("Elements at positions 1 to 3:", arr[1:4])
print("Every second element:", arr[::2])
````

## 6. Vectorized operations

Vectorization lets NumPy apply an operation to each element without an explicit Python loop.

````python
prices = np.array([100, 200, 300])

print("Add 10:", prices + 10)
print("Multiply by 2:", prices * 2)
print("Discounted by 10%:", prices * 0.90)
print("Element-wise addition:", prices + np.array([1, 2, 3]))
````

## 7. Filtering and aggregate functions

````python
marks = np.array([45, 82, 67, 91, 38, 76])

print("Marks above or equal to 70:", marks[marks >= 70])
print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Sorted:", np.sort(marks))
````

## 8. Reshaping arrays

````python
arr = np.arange(1, 7)
matrix = arr.reshape(2, 3)

print(matrix)
print("Shape:", matrix.shape)
print("Flattened:", matrix.flatten())
````

## 9. Data types and type conversion

````python
values = np.array([1, 2, 3])
decimal_values = values.astype(float)

print(values, values.dtype)
print(decimal_values, decimal_values.dtype)
````

## 10. Common interview questions

**Q1. What is an `ndarray`?**  
It is NumPy's core N-dimensional array object.

**Q2. What is vectorization?**  
Applying an operation to an array's elements without writing an explicit Python loop.

**Q3. What is the difference between `shape` and `size`?**  
`shape` gives the dimensions of an array; `size` gives the total number of elements.

**Q4. What is broadcasting?**  
Broadcasting allows NumPy to perform operations on arrays of compatible shapes, sometimes against a scalar or a smaller array.

**Q5. How do you select values greater than 50?**  
Use boolean indexing: `arr[arr > 50]`.

````python
# Broadcasting example
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr + np.array([10, 20, 30]))
````
