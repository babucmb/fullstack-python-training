## 1. What is Pandas?

Pandas is an open-source Python library for working with structured, labelled data. Its two key data structures are:

- **Series:** a one-dimensional labelled array.
- **DataFrame:** a two-dimensional table with labelled rows and columns.

**Interview question: What is the difference between NumPy and Pandas?**

NumPy focuses on numerical arrays and mathematical operations. Pandas provides labelled Series and DataFrames that make tabular work?such as using column names, handling missing values, filtering, grouping, and cleaning?more convenient. Pandas uses NumPy in many of its underlying operations.

```python
import numpy as np
import pandas as pd

print("Pandas version:", pd.__version__)

marks = pd.Series([80, 90, 70], name="marks")
marks
```

## 2. Create a DataFrame

A DataFrame can be created from a dictionary whose values are lists of equal length. Each dictionary key becomes a column name.

```python
data = {
    "name": ["Ravi", "Sita", "John", "Asha", "Mina"],
    "department": ["AIML", "CSE", "CSE", "AIML", "ECE"],
    "marks": [80, 92, 67, 88, 74],
    "attendance": [90, 95, 82, 96, 85],
}

df = pd.DataFrame(data)
df
```

## 3. Inspect data

Inspect a dataset before cleaning or analysing it. These commands show sample rows, dimensions, column names, data types, and non-null counts.

```python
print(df.head())                 # First five rows
print("Shape:", df.shape)        # (rows, columns)
print("Columns:", df.columns.tolist())
print("Data types:", df.dtypes)
df.info()
```

## 4. Select columns and filter rows

Use a single column name to select a Series, a list of names to select a DataFrame, and a Boolean condition to filter rows.

```python
print(df["name"])                         # One column
print(df[["name", "marks"]])              # Multiple columns
print(df[df["marks"] >= 80])               # Marks >= 80
print(df[df["department"] == "AIML"])
```

## 5. Add and update columns

Column calculations are vectorized: the expression is applied to every row without writing a Python loop.

```python
df["passed"] = df["marks"] >= 40
df["bonus_marks"] = df["marks"] + 5
df
```

## 6. Missing values and duplicate rows

Real datasets may contain missing values or duplicated records. Inspect first, then choose a handling strategy based on the data and the task.

```python
messy = pd.DataFrame({
    "name": ["Ravi", "Sita", "Sita", "John"],
    "marks": [80, np.nan, np.nan, 67],
})

print("Missing values per column:")
print(messy.isna().sum())

print("Duplicate rows:")
print(messy.duplicated())

cleaned = messy.drop_duplicates()
cleaned
```

### Filling missing values

For this small example, fill missing marks with the column mean. This is only one possible strategy; the appropriate choice depends on the dataset and task.

```python
filled = messy.copy()
filled["marks"] = filled["marks"].fillna(filled["marks"].mean())
filled
```

## 7. Sorting and summary statistics

Sort rows by a column and use built-in methods to calculate common summary statistics.

```python
print(df.sort_values("marks", ascending=False))
print("Average marks:", df["marks"].mean())
print("Highest marks:", df["marks"].max())
print("Lowest marks:", df["marks"].min())
print("Number of rows:", len(df))
```

## 8. Grouping data with `groupby()`

`groupby()` splits rows by one or more keys, applies a calculation to each group, and combines the results. It is useful for summaries by category.

```python
department_summary = df.groupby("department")["marks"].mean()
print(department_summary)
```

## 11. Common interview questions

**Q1. What is a DataFrame?**  
A two-dimensional labelled data structure with rows and columns.

**Q2. What does `df.isnull().sum()` do?**  
It counts missing values in each column.

**Q3. What does `drop_duplicates()` do?**  
It returns a DataFrame with duplicate rows removed by default.

**Q4. What is `groupby()` used for?**  
It groups rows by category so we can calculate sums, means, counts, and other summaries.

**Q5. How do you improve data quality?**  
Inspect types, missing values, duplicates, invalid values, and inconsistent categories; apply appropriate cleaning rules and validate the output.

**Q6. Why avoid row-by-row loops when possible?**  
Vectorized Pandas operations are often clearer and faster for column-based calculations.
