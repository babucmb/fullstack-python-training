# Comparsion operators (==, >=, <=, !=, <, >)

```python
a=10
b=5

print(a==b)
print(a>=b)
print(a<=b)
print(a!=b)
print(a<b)
print(a>b)
```

# Logical Operators (And , or, not)

```python
# and , or, not

a=True
b=False

print(a and b)
print(a or b)
print(not a)
print(not b)
```

# Identity Operators (is, is not)

```python
a=[10,20,30]
b=[10,20,30]
c=a

print(a==b)
print(a is b)
print(a is c)

print(a is not b)
print(a is not c)
```

#  Membership Operators (in, not in)

```python
# we will use this membership operator to check the value is present in the list or not

fruites=["apple","banana","mango"]
print("apple" in fruites)
print("grapes" not in fruites)
print("banana" not in fruites)
```
