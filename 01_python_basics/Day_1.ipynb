```python
n=18
if n<=1:
    print("not Prime")
else:
    for i in range(2, int(n**0.5)+1):
        if n%i==0:
            print("not prime")
            break
        else:
            print("prime")
            break
```

# Type Casting

# Integer Typecasting

```python
# integer Typecasting : int() function is used to convert the data type into integer.
v1=int(5.6) # float  to integer
v2=int("100") # string to integer

v3=int(True) # boolean to integer
v4=int(False) # boolean to integer
v5=int("1010", 2) # binary to integer

print(v1,type(v1))
print(v2,type(v2))
print(v3,type(v3))
print(v4,type(v4))
print(v5,type(v5))
```

# Float Typecasting

```python
v1=float(5) # integer to float :- Float() function is used to convert the data type into float.
v2=float("10.5") # string to float 
v3=float(True) # boolean to float
v4=float(False) # boolean to float
print(v1,type(v1))
print(v2,type(v2))
print(v3, type(v3))
print(v4, type(v4))
```

# String Typecasting :- str()

```python
v1=str(200) # integer to string :- str() function is used to convert the data type into string.
v2=str(10.5) # float to string
v3=str(True) # boolean to string    
v3=str(False) # boolean to string
print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
```

# Boolean Typecasting

```python
v1=bool(1) # integer to boolean :- bool() function is used to convert the data type into boolean.
v2=bool(0) # integer to boolean
v3=bool(10.5) # float to boolean
v4=bool(0.01) # float to boolean

v5=bool("hello") # string to boolean

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))

print(v5, type(v5))
```

#Q21: A bakery sells custom cookies based on weight. Write a program that reads the price per gram as a decimal and the quantity in grams as an integer. Calculate the final price and display it in the format: Total: <price>. Input Format: The first line contains a float (price), and the second line contains an integer (quantity). Output Format: Print the result as 'Total: ' followed by the calculated value.

```python
price=10.5
quantity=2
total=price*quantity
print(f"Total:",{total})
```
