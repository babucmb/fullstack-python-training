```python
# Example 1: How To define Abstract Class , Abstract Method

from abc import ABC, abstractmethod
# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): #abstract method
        pass
    @abstractmethod
    def LED_Screen(self): #abstract method
        pass
   

class hyd_mahindra(Head_Mahindra): #child class 
    def air_bag(self): #abstract method
        print("I Will use nylon to design airbags")
    def LED_Screen(self):
        print("Sony LED")
    def insurance(self): # Concrete
        print("it is your option to take  insurance @ hyd_m ")
class noida_mahindra(Head_Mahindra):
     def air_bag(self): #abstract method    
        print("I Will use cotton to design airbags")
     def LED_Screen(self):
        print("Samsung LED")
     def insurance(self): # Concrete
        print("Please take Mahindra company insurance")        
obj_hyd = hyd_mahindra()
obj_noida = noida_mahindra()
# obj_hyd.insurance()
obj_noida.insurance()
#obj_head = Head_Mahindra()
```
