# Encapsulation

Def-> Encapsulation refers to the binding of data(attributs) and function into one unit

--> Encapsulation restrict the drict access of the data
--> We can access the varible/data/attribute/ thrugh object outside the class.
--> We can access the varible/data/attribute through functions defined inside the class.

```python

class bike:
    no=2 # class variable/ Attribute/data
    def __init__(self, name, price):
        self.bike_name=name # instance variable
        self.bike_price=price
        print(self.bike_name, end=" ")
        print(self.bike_price)
    
    # Creating member function
    def dis(self):
        print(self.no,self.bike_name,self.bike_price)
#obj1=bike("FZ",10000000)
#obj2=bike("TVS",2000000)
obj1=bike("RE",1.5)
obj1.dis()
print(obj1.no)
print(obj1.bike_name)

print(obj1.bike_price)
# Calling the Memeber function
# obj1.dis()
```

# Access specifers / type of attributes in encapsulation

Types:-
--> 1. Public:- These varibles can be accessed inside/outside/child the class
            -->> Public varibles Don't start with Underscore
--> 2. Protected:- These varibles can be accessed only in inside/child the class
            -->> Protected varible NOT INTENDED to access outside the class
            -->> Protected varible start with SINGLE underscore
            -->> Procted variables can be accessed outside this class, but you should not access.
            -->> In python public, protected, private concept is used just for Sign/warning
--> 3. Private:- These varibles can be accessed only within inside the class.
            -->> Not INTENDED to access outside/child class
            -->> Private vaibles start with DOUBLE Underscore


```python
class marvels:
    def __init__(self, name, power, weapon): # self will indictes that calling the construtuor
        self.Name=name # public instance variable
        self._Power=power # Protected instance variable
        self.__Ayudham=weapon # Private instance variable
    def benten(self): 
        print(self.Name, self._Power, self.__Ayudham)
obj1=marvels("Groot", 100000, "Tree Hand")
obj1.benten()

print(obj1.Name)
print(obj1._Power) # --> Procted variables can be accessed outside this class, but you should not access.
# --> In python public, protected, private concept is used just for Sign/warning
# print(obj1.__Ayudham) ---> it is private variable, can't access the outside the class
# But by using the class name we can access the private varible here:-
print(obj1._marvels__Ayudham)
```

# Types of Inheritance:
1. single Inheritance:- One parent - one child
2. Multiple Inheritance:- Many parents - one child
3. mutilevel Inheritance:- child-> father-> grand father-> forfather
4. Hierarical Inheritance:- one parent - many child
5. Hybrid Inheritance:- many parent - many child . combine any two or more Inheritance




# 1. Single Inheritance:- One parent - One child

```python
# Example-1 Structure of singel inheritance
# Parent Class
class hyd_food:
    taste="Very Spicy"

    def biryani(self):
        print("Yes it is very Spicy")
# child class
class kfc(hyd_food):
    pass
obj_child=kfc()
print(obj_child.taste)
obj_child.biryani()
```

```python
# other example
class animal():
    def make_sound(self):
        print("animal making sound")
class dog(animal):
    def barking(self):
        print("Dog is barking")

obj1=dog()
obj1.barking()
obj1.make_sound()
```

```python
# Example -2 super() function

#super(): used to access the parent properties and function from inside of the child class

class hyd_food:
    taste="Very Spicy"

    def biryani(self):
        print("Yes it is very Spicy")
# child class
class kfc(hyd_food):
    def display(self):
        super().biryani()
        print(self.taste)

obj_child=kfc() # it is the object of child class
print(obj_child.taste)
obj_child.biryani()
```

```python
class hyd_food:
    taste="Very Spicy"

    def __init__(self):
        print("Yes it is very Spicy")
# child class
class kfc(hyd_food):
    def __init__(self):
        print("I am KFC King")

obj_child=kfc() 
obj_child=hyd_food()
obj_child.taste
# if the class constructor function is not defiend
#  then child class object will call automatically parent class function
```

# 2. Multiple Inheritance:- Many parents - one child

```python
class animal():
    def make_sound(self):
        print("animal making sound")
class dog():
    def barking(self):
        print("Dog is barking")

class puppy(dog,animal):
    def baby(self):
        print("puppy sounnding")

obj1=puppy()
obj1.baby()
obj1.barking()
obj1.make_sound()
```

```python
# Example-1: Structure for Mutiple inheriteance
class A:
    def display(self):
        print("I am a A class")
class B:
    def show(self):
        print("I am B class")
class C(A,B):
    pass
obj1=C()
obj1.display()
obj1.show()
```
