# Multilevel inheritance

```python
class granter:
    def Gold(self):
        print("1000 KG")

class grandfather(granter):
    def Land(self):
        print("500 Acr")
        
class father(grandfather):
    def car(self):
        print("Lambo")
class son(father):
    def Bike(self):
        print("RE Bike")
obj_son = son()
obj_son.Bike()
obj_son.car()
obj_son.Land()
obj_son.Gold()
```

# Hydrid Inheritance

```python
class A:
    A_var = "A Class"
class B(A):
    B_var = "B Class"
class C(B):
    C_var = "C Class"
class D(B):
    D_var = "D Class"
class E(C,D):
    E_var = "E Class"
obj_E = E()
print(obj_E.A_var)
```

```python
# Example :-2 Constructor
class A:
    def __init__(self):
        print("A Class")
    A_var="A Class"  
class B(A):
    def __init__(self):
        print("B Class")
    B_var = "B Class"
class C(B):
    def __init__(self):
        print("C Class")
    C_var = "C Class"
class D(B):
    def __init__(self):
        print("D class")
    D_var = "D Class"
class E(C,D):
    E_var = "E Class"
obj_E = E()
print(obj_E.A_var)
```

```python
# Example :-3 Super()
class A:
    A_var = "A Class"
    def A_output(self):
        print("This is  A Output")
class B(A):
    B_var = "B Class"
    def B_output(self):
            print("This is B Output")
class C(B):
    C_var = "C Class"
    def C_output(self):
            print("This is  C Output")
class D(B):
    D_var = "D Class"
    def D_output(self):
            print("This is D Output")
class E(C,D):
    E_var = "E Class"
    def E_output(self):
            print("This is E Output")
obj_E = E()
print(obj_E.A_var)
print([i.__name__ for i in E.__mro__])
```
