# Polymorphism

--> Ploy means "Many form"
--> In ploymorphism, same fucntion or same operator can perform many different task.
--> Types of Ploymorphism
    1. Function overloading
    2. Operator overloading
    3. Method overriding


# 1. Function overloading

```python
# This type function overloading is supported in C++ and Java
# Based upon number of arguments passed to funtion, it will handles the function overloading

def fun(var1):
    print(var1)
def fun(var1,var2):
    print(var1,var2)
fun(10)
```

# Using Default Arguments for function overloading


```python
# Example-1
def fun(v1,v2=None,v3=None):
    if v2 is not None:
        print(v1,v2,v3)
    else:
        print(v1)
fun(10)
fun(10,20)
fun(10,20,30)
```

```python
# Example-2:
def cal(a=0,b=0,c=0,d=0):
    return a+b+c+d
print(cal(10,20))
print(cal(10,20,30))
print(cal(10,20,30,40))
print(cal())
```

#   *args for function overloading

```python
# Example-2:
def cal(*args):
    return sum(args)
print(cal(10,20))
print(cal(10,20,30))
print(cal(10,20,30,40,50,60,70,80,0,100))
print(cal())
```

```python
# Function overloading with decorator
from multipledispatch import dispatch
@dispatch(int)
def fun(v1):
    print(v1)
@dispatch(int,int)
def  fun(v1,v2):
    print(v1,v2)

@dispatch(float,float)
def  fun(v1,v2):
    print(v1,v2)
fun(10,20)
fun(1.1,2.1)
```

# Operator Overloading

--> We will perform different operationson the user defined objects using operators.

```python
class D:
    def __init__(self,v,p):
        self.v=v
        self.p=p

    def __add__(self,d2_val):
        tp=self.p+d2_val.p
        return tp
    def __sub__(self,d3_val):
        tp=self.p-d3_val.p
        return tp
    def __mul__(self,d4_val):
            tp=self.p*d4_val.p
            return tp
    def __truediv__(self,d5_val):
            tp=self.p/d5_val.p
            return tp
    def __mod__(self,d6_val):
            tp=self.p%d6_val.p
            return tp
d1=D("Masala",60)
d2=D("Onine",70)
print(d1+d2)
print(d1-d2)
print(d1*d2)
print(d1/d2)
print(d1%d2)
# print(d1**d2)  __pow__ function for power
```
