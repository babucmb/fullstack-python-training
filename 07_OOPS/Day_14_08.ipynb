```python
# Example-1: Structure of Class

# Class is the keyword in python, food is the name of class
# class variable :- The value of class variable remains SAME for all the objects
# class varible value cannot be change by object
# class varible value CAN be only change by class name
class food:
    food_name="Biryani" # class Variable

obj3=food()
obj3.food_name="papu"  # here obj3 is creating its own local varibale with same name
print(obj3.food_name)  # printing Local variable to obj3


obj=food() # creating the object (obj) of food() class
print(obj.food_name) # accessing the class varible

food.food_name="Mutton"

obj2=food()
print(obj2.food_name) # class object obj2 can acces the class data

# access the class variable with the name
print(food.food_name)

# Class varible value cannot be channge by object
'''obj3=food()
obj3.food_name="papu"  # here obj3 is creating its own local varibale with same name
print(obj3.food_name)  # printing Local variable to obj3'''
```

```python
# Example-2 :- Constructure function
# Constructor function automatically get called when object is created.
# Constructor is created by __init__
# Constructor  function is used for intializing the varaiable

# instance variable value id DIFFERENT for every object.
# class varaible value is SAME for all the object

# self is a varaible which points to the current object

class bike:
    no=2 # class variable
    def __init__(self, name, price):
        self.bike_name=name # instance variable
        self.bike_price=price
        print(self.bike_name, end=" ")
        print(self.bike_price)
obj1=bike("FZ",10000000)
obj2=bike("TVS",2000000)
```

```python
# Example-3:- Member Functions

# Member function don't get called automatically, we need to call them our self

class bike:
    no=2 # class variable
    def __init__(self, name, price):
        self.bike_name=name # instance variable
        self.bike_price=price
        print(self.bike_name, end=" ")
        print(self.bike_price)
    
    # Creating member function
    def dis(self):
        print("I am a Member Function")
obj1=bike("FZ",10000000)
obj2=bike("TVS",2000000)

# Calling the Memeber function
obj1.dis()
```
