# Creating Different Types of Tuples

```python
em=()
em2=tuple()
print(em,type(em))
print(em2,type(em2))
```

```python
het_tuple=(1,2,"Mahesh", True, 3.5)
print(het_tuple,type(het_tuple))
```

```python
# Duplicates value tuple
dup_tuple=(1,2,3,4,5,1,2,3)
print(dup_tuple,type(dup_tuple))
```

```python
dup_tuple=(1,2,[1,2,3],3,4,{"mahesh","adi","krishna"},5,1,{"I":10,"II":20},2,3)
print(dup_tuple,type(dup_tuple))
dup_tuple[2].append(99)
print(dup_tuple)
my_set = list(dup_tuple[5])
print(my_set[1])
print(my_set)
```

```python
# convert list to tuple
list1=[1,2,3,4,5]   
tuple1=tuple(list1)
print(tuple1,type(tuple1))
list2=list(tuple1)
```

```python
val=range(1,10)
tu=tuple(val)
print(tu,type(tu))
print(tuple(range(1,10,2)))
```

```python
t1=(1,2,3,4,5)
t2=(6,7,8,9,10)
t3=(11,12,13,14,15)
t4=t1+t2+t3
print(t4,type(t4))
print(t1+t2+t3)
t5=tuple(i for i in range(10))
print(t5,type(t5))
```

# Accessing the Tuple Items


```python
veg_tabel=("Tomato","Potato","Brinjal","Ladyfinger","Cabbage")
print(veg_tabel,type(veg_tabel))
veg_tabel[-1:0:-1]
print(veg_tabel[:])
```

```python
tuple1=(10,20,10,20.5,[1,2,3], True)
print(tuple1,type(tuple1))
tuple1[4][0]=500
#tuple1[2].append(99)  # This will raise an error because tuple elements are immutable
print(tuple1)
```

```python
# Tuple item can't update directly but we can update the mutable item inside the tuple.
# complete item in tuple cannot be updated directly.
# but if the item is mutable like list, dictionary, set then we can update the mutable item inside the tuple.
```

```python
tuple1=(1,2,3,4,5)
list1=list(tuple1)
print(list1,type(list1))

```

# Popular inbuilt Funcation for Tuple

```python
salary=(10000,20000,30000,60000,50000)
print(len(salary))
print(max(salary))
print(min(salary))
print(sum(salary))
print(sorted(salary))
# sort() is inplace funcation
# inplace means sorting is done in the same data structure and it will not return any value. It will return None.

# sorted() funcation return the list
# sort() funcation don't return anything.
```
