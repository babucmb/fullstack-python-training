# Exception Handling

--> Def:- Exception Handling allows to respond to error without crashing the running application
--> With Exception Handling, we can give our own error message

--> Types Of Blocks in Exception Handling:
1. Try:- put that code where error can our
2. Expect:- Handle the error
3. Else:- put code which you want to execute when their is No error
4. Finally:-execute either error or No error


--> Different types Error in Exception error
1. NameError:- Variable is not defined
2. IndexError
3. ZeroDevisionError
4. ArrayOutOfIndexError
5......- like that any types of error there over inn python

```python
# Example:-1
print(x)
print("Hello")
a,b=10,20
print(a+b)
```

```python
# Example 2:- Structure of try-except

try:
    print(x)
except Exception as e:
    print(f"compiler error:{e}")
    print(type(e))
print("Hello")
a,b=10,20
print(a+b)
```

```python
# Example 3:- else Block

try:
    x=10
    print(x)
except Exception as e:
    print(f"compiler error:{e}")
    print(type(e))
else:
    print("No error in try Block")
```

```python
# Example 4:- finally Block

try:
    x=10
    print(z)
except Exception as e:
    print(f"compiler error:{e}")
    print(type(e))
else:
    print("No error in try Block")
finally:
    print("I will execute whatever happen")
```

# Types of Errors

```python
# NameError:-

try:
    print(y)
except Exception as e:
    print("This is name error")
    print(type(e))
```

```python
# ZeroDivisionError:-

try:
    y=10
    print(y/0)
except Exception as e:
    print("This is error")
    print(type(e))
```

```python
# NameError:-
try:
    print(z)
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is Zero Division Error")
```

```python
# ZeroDivisionError:-

try:
    y=10
    print(y/0)
except NameError:
    print("This is Name erro")
except ZeroDivisionError:
    print("This is Zero Division erro")
```

```python
# IndexError:-

try:
    l1=[10,20,30]
    print(l1[5])
except NameError:
    print("This is Name erro")
except ZeroDivisionError:
    print("This is Zero Division erro")
except IndexError:
    print("This is Index error")
except Exception as e:
    print(type(e),e)
```

```python
try:
    d1={"name":"Adhi"}
    print(d1["age"])
except NameError:
    print("This is Name erro")
except ZeroDivisionError:
    print("This is Zero Division erro")
except IndexError:
    print("This is Index error")
except KeyError:
    print("This is KeyError")
except Exception as e:
    print(type(e),e)
```
