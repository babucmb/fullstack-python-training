# For Loop

```python
an=["puli",  "pili", "kukka", "pandi", "kothi"]
for i in an:
    print(i)
```

```python
fdn="chicken food"
for i in fdn:
    print(i)
```

```python
# For loop with range()
# range() funcation is used to generate list of elements
# Syntax of range:- range(start, end, step)
# start=00
#end=end-1
# step =1

for i in range(2,10):
    print(i)
```

```python
for i in range(2,10,3):
    print(i)
```

```python
# example-4: Break Statement
# Break Statement wil bring the control out of loop

for num in range(1,6):
    if num==3:
        break
    print(num)
```

```python
#  Example -5 Continue statement
# Continue statement take the control to the start of the loop

for num in range(1,6):
    if num==3:
        continue
    print(num)
```

```python
# Example -6 : Nested Loop
for i in range(1,4):
    for j in range(1,3):
        print(i,j)
```

```python
# Example-6:- Nested Loop
# For every value of outer Loop, inner loop will completely execute
for i in range(3,6): # n=3 times
    for j in range(2,7):  # m=5 times
        print(i,j)
# O(n*m)- time Complexity
```

```python
# Exampe-7 : else Statement
# else statement executes if loop complete Execution successfully

for i in range(5):
    if i==2:
        continue
    elif i==4:
        break
    else:
        print(i)
    pass
else:
    print("Loop completed Successfuly")
```

```python
c=5
r=5
for i in range(c):
    for j in range(5):
        print("*", end="")
    print()
```

```python
r=5
for i in range(r):
    for j in range(i+1):
        print("*", end=" ")
    print()
```

```python
r=5
for i in range(r,0,-1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
r=5
for i in range(1,r+1):
    for j in range(r-i):
        print(" ", end="")

    for k in range(2*i-1):
        print("*", end="")
    print()
```
