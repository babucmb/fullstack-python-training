# Break Statment

```python
# Example -2 :- Break Statment
# Break statement helps too come out of the loop

i=1
while (i<=10):
    print(i)
    if i==5:
        break
    i+=1
```

# Countinue Statement

```python
# Example-3
i=1
while i< 10:
    i+=1
    if i < 5:
        continue
    if i> 8:
        break
    print(i)
```

```python
# Example -4 else statement with while loop
# else statement executes only if loop is successfully completed

i=1
while i<=10:
    print(i)
    i+=1
    if i>5:
        break
```
