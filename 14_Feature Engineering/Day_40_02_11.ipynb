# 1. Label encoding


```python
# Def:- Assign a
```

```python
# Example-1
from  sklearn.preprocessing import LabelEncoder
import pandas as pd

data = pd.DataFrame({'color':['red','green','blue','green','red']})
label_encoder = LabelEncoder()
# fit() fuction allot the interger value to each unique value of the categorical data
# transform() function convert the categorical data to numerical data
data['color_encoded'] = label_encoder.fit_transform(data['color'])
print(data)
```

# 3. Ordinal encoding

```python
# Def:- Oridnal encoding converts the data into numerical data based upo the ranking of the data. It is used when the categorical data has some order or ranking.

from sklearn.preprocessing import OrdinalEncoder
data = pd.DataFrame({'size':['small','medium','large','medium','small']})
ordinal_encoder = OrdinalEncoder(categories=[['small','medium','large']])
data['size_encoded'] = ordinal_encoder.fit_transform(data[['size']])
print(data)
```

# 2. One hot encoder

# Method 1:-

```python
# Converts categorical data into binary indicators.
# Drawback:- It increases the number of features in the dataset. It is not suitable for high cardinality categorical data. memory consumption is high.
# Don't use one hot encoding for high cardinality categorical data. Use label encoding or ordinal encoding.
# Use one hot encoding for low cardinality categorical data.

data=pd.DataFrame({'Country':['india','Thailand','USA','japan','Nigeria']})
encoded_data=pd.get_dummies(data,columns=['Country'],dtype=int)
# encoded_data=pd.get_dummies(data,dtype=int)
print(encoded_data)
```

# Method-2:-

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False,dtype=int)
encoded_data = encoder.fit_transform(data[['Country']])
print(encoded_data)
```
