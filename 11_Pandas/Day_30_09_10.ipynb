```python
# Approach1 :- delete the records which having null values
# Approach2:- delete the column which having null values

df2=df.copy()
```

```python
# axis=1:- delete the columnn which is having null values
# axis=0:- delete the row which is having null values
# how='any' :- if any null value is there then delete the column
# how='all' :- if all values are null then delete the column
df2=df2.dropna(axis=1, how='any')
df2.isnull
```

```python
df2=df2.dropna(axis=1, how='all')
df2.isnull().sum()
```

```python
df3=df.copy()   
```

```python
# fillna() ffunction is used to fill the null values with some other value
df3.Age=df3.Age.fillna(18)
```

# Handling the Duplicates Values

```python
# Identify the duplicates records
df.duplicated().sum()
```

```python
# show the duplicate records
df[df.duplicated()]
```

```python
# keep=False will show all the duplicate records
# keep=true will show only the last duplicate record
df[df.duplicated(keep=False)]
```

```python
df[df.duplicated(keep='last')]
```

```python
# deleting the duplicate records
df=df.drop_duplicates()
```
