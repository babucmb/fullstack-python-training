# Data Transformation using apply() function & map() function

```python
# apply() : this functionis used to excute lambda function 
# Lambda function is one line function
# if you want to apply lambda function on a column then you can use apply() function
```

```python
df1=df.copy()
```

```python
# on the festival occation : sai ram decide to double the employee salary
df
```

```python
df1.salary=df1.salary.apply(lambda x:x*2)
df1
```

```python
# Create a new column Bonus or NO Bonus
df1["Diwali_gift"]=df1.apply(lambda row: "Bonus" if row['salary']> 100000 else "No Bonus", axis=1)
```

```python
df1
```

```python
# Create a new column Bonus or NO Bonus
# apply() function is used to apply the lambda function on the row of the dataframe
df1["Diwali_gift"]=df1.apply(lambda row: "Bonus" if (row['salary']<300000 and row['Age']> 40) else "No Bonus", axis=1)
```
