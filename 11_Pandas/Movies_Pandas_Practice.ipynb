# Movies Dataset: Pandas Practice Queries

This notebook answers 20 Pandas practice queries using `Movies_Dataset.xlsx`. Each section is independent and can be run after the setup cell.

## Setup and data loading

The workbook contains the columns `Movie Name`, `Genre`, `Release_Year`, `IMDb_Rating`, and `Revenue`. Revenue values are treated as the dataset's box-office collection measure.

```python
from pathlib import Path
import pandas as pd

DATA_FILE = Path(r"C:\Users\MAHESH\Desktop\fullstack-python-training\Movies_Dataset.xlsx")
df = pd.read_excel(DATA_FILE)

# Keep the original data unchanged; later transformation queries use copies.
df.head()
```

```python
# Quick inspection
print(f"Rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
df.info()
```

## 1. Select all movies in the Action genre

```python
action_movies = df[df["Genre"].eq("Action")]
action_movies
```

## 2. Select movies released after 2000

```python
movies_after_2000 = df[df["Release_Year"] > 2000]
movies_after_2000
```

## 3. Select movies with an IMDb rating greater than 8.5

```python
high_rated_movies = df[df["IMDb_Rating"] > 8.5]
high_rated_movies
```

## 4. Select only Movie Name and Genre columns

```python
movie_names_and_genres = df[["Movie Name", "Genre"]]
movie_names_and_genres
```

## 5. Select movies released between 1990 and 2010

```python
movies_1990_to_2010 = df[df["Release_Year"].between(1990, 2010)]
movies_1990_to_2010
```

## 6. Count the number of movies in each genre

```python
movies_per_genre = df["Genre"].value_counts().rename("Movie Count")
movies_per_genre
```

## 7. Find the highest IMDb-rated movie

```python
highest_rated_movie = df.loc[[df["IMDb_Rating"].idxmax()]]
highest_rated_movie
```

## 8. Find the average IMDb rating of all movies

```python
average_imdb_rating = df["IMDb_Rating"].mean()
print(f"Average IMDb rating: {average_imdb_rating:.2f}")
```

## 9. Find the total box-office collection of all movies

```python
total_box_office = df["Revenue"].sum()
print(f"Total box-office collection: {total_box_office:,.2f}")
```

## 10. Select movies with box-office earnings greater than 500 million

```python
movies_over_500_million = df[df["Revenue"] > 500]
movies_over_500_million
```

## 11. Rank movies based on IMDb rating

```python
ranked_movies = df.assign(
    IMDb_Rank=df["IMDb_Rating"].rank(method="dense", ascending=False).astype(int)
).sort_values(["IMDb_Rank", "Movie Name"])
ranked_movies[["IMDb_Rank", "Movie Name", "IMDb_Rating", "Genre"]]
```

## 12. Sort movies by release year in descending order

```python
movies_by_latest_year = df.sort_values("Release_Year", ascending=False)
movies_by_latest_year
```

## 13. Find all duplicate movie names, if any

```python
duplicate_movie_names = df[df["Movie Name"].duplicated(keep=False)].sort_values("Movie Name")
duplicate_movie_names
```

## 14. Remove duplicate movie names

```python
movies_without_duplicate_names = df.drop_duplicates(subset="Movie Name", keep="first").reset_index(drop=True)
print(f"Rows before: {len(df):,}")
print(f"Rows after removing duplicate names: {len(movies_without_duplicate_names):,}")
movies_without_duplicate_names
```

## 15. Replace Sci-Fi genre with Science Fiction

```python
movies_with_updated_genre = df.copy()
movies_with_updated_genre["Genre"] = movies_with_updated_genre["Genre"].replace({"Sci-Fi": "Science Fiction"})
movies_with_updated_genre
```

## 16. Select the first 10 movies

```python
first_10_movies = df.head(10)
first_10_movies
```

## 17. Select the last 5 movies

```python
last_5_movies = df.tail(5)
last_5_movies
```

## 18. Group movies by genre and find the average IMDb rating for each genre

```python
average_rating_by_genre = (
    df.groupby("Genre", as_index=False)["IMDb_Rating"]
      .mean()
      .rename(columns={"IMDb_Rating": "Average IMDb Rating"})
      .sort_values("Average IMDb Rating", ascending=False)
)
average_rating_by_genre
```

## 19. Find the number of movies released in each year

```python
movies_per_year = (
    df.groupby("Release_Year")
      .size()
      .rename("Movie Count")
      .reset_index()
      .sort_values("Release_Year")
)
movies_per_year
```

## 20. Find the total box-office collection per genre

```python
box_office_by_genre = (
    df.groupby("Genre", as_index=False)["Revenue"]
      .sum()
      .rename(columns={"Revenue": "Total Box Office Collection"})
      .sort_values("Total Box Office Collection", ascending=False)
)
box_office_by_genre
```
