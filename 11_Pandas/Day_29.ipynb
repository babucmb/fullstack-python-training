# ** Filtering & Idexing  

```python
# Query : find those records where salary >100000
df[(df.salary > 50000) & (df.salary < 100000)]
```

```python
# Query : find those records where salary < 500000 or salary >150000 
df[(df.salary < 50000) | (df.salary < 100000)]
```

```python
# Query : find max salary 
df[df.salary == df.salary.max()]
# max salary 
df[df.Marriage_proposal_count == df.Marriage_proposal_count.max()]
df[df.salary == df.salary.min()][['Name', 'Profession']]
df.loc[df['salary'] == df['salary'].min()][['Name','Profession']]
```

# ** Indexing

```python
# loc -> filtering fixed number of indexes
df.loc[1:6, 'Name':'Profession']
```

```python
df.loc[10 : 20, 'Age':'salary']
df.loc[10 : 20 : 2, ['Age','salary']]
```

```python
# iloc -> index based fetching

# Query : printrecord from 1 to 6 and columns from 1 to 3
# df.iloc[row,col]
df.iloc[1:7, 1:4]
```

```python
# Interv Questions

# Diff b/w loc & iloc
# loc we need to mention the index of columns
# while in iloc we mention index of columns

# iloc goes upto n-1, but loc goes upto nth value

# iloc is little faster than loc bec loc needs to convert colmns to index
# 
```
