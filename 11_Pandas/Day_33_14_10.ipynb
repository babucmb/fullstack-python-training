# Time series based Dataframe

```python
import pandas as pd
```

```python
date_rng=pd.date_range(start='1/1/2020', end='1/08/2020', freq='h')
date_rng
```

```python
import numpy as np
# choose raandom 100 data values from the date_rng
# random.choice() function is used to choose random values from a given array.
# replcae=False: dont choose te same value again and again
random_dates = np.random.choice(date_rng, size=100)
random_dates
```

```python
# Time series based dataframe: dataframe with datae/time column as index is called time series based dataframe.
# pd.to_datetime() : converting numpy date_time data into pandas date_time data type
#np.arange(start,end): generates a sequence of numbers from start to end-1
# np.random.randint(): generating random intergers: 20 to 199


df=pd.DataFrame({
    'timestamp':random_dates,
    'oeder_id':np.arange(1,101),
    'catrgory_food':np.random.choice(['veg','non-veg'],size=100),
    'price': np.random.randint(20,200,size=100),
    'quity': np.random.randint(1,10,size=100)
})
df['year'] = df['timestamp'].dt.year
df['month'] = df['timestamp'].dt.month
df['days'] = df['timestamp'].dt.day

df['hour'] = df['timestamp'].dt.hour
df['minute'] = df['timestamp'].dt.minute
df['second'] = df['timestamp'].dt.second

df
```

# Chart building using pandas

```python
df.groupby('year').quity.sum().plot(kind='bar',title='year wise quity sold')
```

```python
df.groupby('year').quity.sum().plot(kind='pie',title='year wise quity sold')
```

```python
df.groupby('year').quity.sum().plot(kind='line',title='year wise quity sold')
```
