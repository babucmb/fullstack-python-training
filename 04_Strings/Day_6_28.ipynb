# 3. Membership (in, Not in)

```python
st1="Hello world"
print("Hello" in st1)
print("Helll" in st1)
print("o W" in st1)
print("Helo" not in st1)
print("Hello" not in st1)
print(len(st1))
```

```python
# string [start:end:step]
h="Samantha"
print(h[4])
print(h[-6])
print(h[0:5])
print(h[0:5:2])
print(h[:3]+ h[-2:])
print(h[::])
print(h[6:3:-1])
```

```python
h="samantha"
c=""
if r in h:
    c=r+c
print(c)
```

```python
st="mahesh"
r=""
for ch in st:
    r=ch +r
print(r)
```
