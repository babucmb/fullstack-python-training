# Conditional Statements in Python (if, else, elif)

```python
# Explain The if Statement in Python, The `if` statement in Python is a control flow statement that allows you to execute a block of code only if a specified condition is true. It is used to make decisions in your code based on certain conditions.
a=10
if a%2==0:
    print(a)
```

```python
# example 2 if-else condition, it is used to execute one block of code if the condition is true, and another block of code if the condition is false.

a=9
if a%2==0:
    print("EVEN")
else:
    print("ODD")
```

```python
# Teranary conditon:- wrting a if - else statement in a single line is called teranary condition. It is used to execute one block of code if the condition is true, and another block of code if the condition is false.
print("EVEN") if a%2==0 else print("ODD")
```

```python
if a%3==0:
    print("Hello")
else:
    print("world")
```

```python
# elif Condition:- The `elif` statement in Python is used to check multiple conditions in a sequence. It allows you to specify additional conditions to be checked if the previous conditions are not met. The `elif` statement is short for "else if" and is used in conjunction with the `if` statement.
b=input("Famous Biryani")
if b=="Paradise":
    print("Spicy & Tasty")
elif b=="Shahdab":
    print("Double Masala")
elif b=="road size":
    print("Budget Friendly")
else:
    print("PG Biryani")
```

```python
# replace elif with else: if, it is about block creations.
b=input("Famous Biryani")
if b=="Paradise":
    print("Spicy & Tasty")
else:
    if b=="shahdab":
        print("Double Masala")
    else:
        if b=="road sixe":
            print("Budget friendly")
        else:
            print("PG Biryani")
```

# Match Case

```python
# Match case is used to replace multiple if else condition.
# Match case is more powerful than the switch case statment.
# Match cse works with varible and data structures.
```

```python
# Explame-1
num=int(input("enter Number"))
match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print("you entered 30")
    case _:
        print("something else")
```

```python
#  Explame -3 : if condition with match case

val=int(input("Enter number"))
match val:
    case x if x>0:
        print("It is a positive NB")
    case y if y<0:
        print("It is a Negative NB")
    case _:
        print("It is a zero")
```
