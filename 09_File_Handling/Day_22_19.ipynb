# Handling

```python
import os 
os.getcwd()
```

```python
# open("file_name","mode")
# f is file object which we can access the content of file
f=open("File_handling.txt","r")
for line in f:
    print(line)
f.close()
```

```python
# with statement:- this will automatically close the file
with open("File_handling.txt","r") as f:
    for line in f:
        print(line)
```

```python
# read(): to get all lines  from the file
with open("File_handling.txt","r")  as f:
    print(f.read())
```

```python
# readlines: to get all the lines into a list, to access particular line  data
with open("File_handling.txt","r") as f:
    lines=f.readlines() 
    print(lines)
```

```python
with open("File_handling.txt","r") as f:
    lines=f.readlines() 
    print(*lines)
```

# Write() Function

--> To writee the content inside the file
--> we will use the "w" mode with write() function
--> if file dont exist then "w" mode will create create a new file.
--> read mode "r" dont create the file

```python
with open("File__handlin_2.txt","w") as f:
    f.write("Adhi is bad girl")
    
```

```python
with open("File__handlin_2.txt","r") as f:
    print(f.read())
```

# Append mode :a

```python
with open("File__handlin_2.txt","a") as f:
    f.write("Ajith is a bad boy")
    
```

```python
with open("File__handlin_2.txt","r") as f:
    print(f.read())
```

# Writelines: Writing multiple line together in the file

```python
with open("File__handlin_2.txt","a") as f:
    f.writelines(["Rohith is friend of adhi. \n","Still Bhavith is also friend of rohith"])
```

```python
with open("File__handlin_2.txt","r") as f:
    print(f.read())
```
